## PANDAS COMMANDS
    1. read_csv() : read csv file
    2. head() : display first n rows of the dataframe
    3. tail() : display last n rows of the dataframe
    4. info() : display information about the dataframe
    5. describe() : generate descriptive statistics of the dataframe
    6. shape : return the shape of the dataframe
    7. columns : return the column names of the dataframe
    8. dtypes : return the data types of the columns in the dataframe
    9. isnull() : check for missing values in the dataframe
    10. dropna() : remove rows with missing values
    11. fillna() : fill missing values with a specified value
    12. drop_duplicates() : remove duplicate rows from the dataframe
    13. duplicated() : check for duplicate rows in the dataframe
    14. value_counts() : count the number of occurrences of each unique value in a column
