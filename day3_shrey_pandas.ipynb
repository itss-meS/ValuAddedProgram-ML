## PANDAS :
    1. PANDAS IS A LIBRARY BUILT ON TOP OF NUMPY.
    2. IT PROVIDES DATA STRUCTURES AND DATA ANALYSIS TOOLS.
    3. IT IS USED FOR DATA MANIPULATION AND ANALYSIS.
    4. IT PROVIDES TWO MAIN DATA STRUCTURES : SERIES AND DATAFRAME.

```python
import pandas as pd
import numpy as np
print("pandas version:", pd.__version__)
```

## PANDAS SERIES :
    1. A SERIES IS A ONE-DIMENSIONAL LABELED ARRAY CAPABLE OF HOLDING ANY DATA TYPE.
    2. IT CAN BE CREATED FROM A LIST, DICTIONARY, OR NUMPY ARRAY.
    3. IT HAS AN INDEX THAT LABELS EACH ELEMENT IN THE SERIES.

```python
#  create a pandas series
s = pd.Series([1, 2, 3, 4, 5])
print(s)
```

```python
# pandas series with empty list
s_empty = pd.Series([])
print(s_empty)
```

```python
## pandas series from list
s_from_list = pd.Series(['a', 'b', 'c', 'd', 'e'])
print(s_from_list)
```

```python
## using numpy array to create a pandas series
arr = np.array([10, 20, 30, 40, 50])
s_from_array = pd.Series(arr)
print(s_from_array)
```

```python
# properties of pandas series
s = pd.Series([1, 2, 3, 4, 5])
print("Index:", s.index)
print("Values:", s.values)
print("Data type:", s.dtype)
print("Length:", len(s))
```

```python
## pandas series from list
s_from_list = pd.Series(['a', 'b', 'c', 'd', 'e',9.8])
print(s_from_list)
print("dtype:", s_from_list.dtype)
```

## PANDAS DATAFRAME :
    1. A DATAFRAME IS A TWO-DIMENSIONAL LABELED DATA STRUCTURE WITH COLUMNS OF POTENTIALLY DIFFERENT TYPES.
    2. IT CAN BE THOUGHT OF AS A TABLE OR A SPREADSHEET.
    3. IT CAN BE CREATED FROM A DICTIONARY OF LISTS, A LIST OF DICTIONARIES, OR A NUMPY ARRAY.

```python
# Create a pandas dataframe
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df = pd.DataFrame(data)
print(df)
```
