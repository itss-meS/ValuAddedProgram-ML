## simple random sampling

```python
import numpy as np
import pandas as pd
population = np.arange(1, 101)  # Population of numbers from 1 to 100
sample = np.random.choice(population, size=10, replace=False)
sample_df = pd.DataFrame({'sample': sample})
print(sample_df)
```

```python
df = pd.DataFrame({'Age': np.arange(1, 101), 'Income': np.random.rand(100)})
sample_random = df.sample(n=10, random_state=42)
print(sample_random)
```

## systematic random sampling

```python
population = np.arange(1, 101)  # Population of numbers from 1 to 100
k = 10  # Sample size
start = np.random.randint(0, k)  # Random starting point
systematic_sample = population[start::k]  # Select every k-th element
systematic_sample_df = pd.DataFrame({'systematic_sample': systematic_sample})
print(systematic_sample_df)
```

## stratafied random sampling

```python
df = pd.DataFrame({
    'Age': np.random.randint(18, 60, size=100),
    'Income': np.random.rand(100),
    'Gender': np.random.choice(['Male', 'Female'], size=100)
})
stratified_sample = df.groupby('Gender', group_keys=False).apply(lambda x: x.sample(frac=0.1, random_state=42))
print(stratified_sample)
```
