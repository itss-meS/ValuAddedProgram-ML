```python
# Question 03: Which built-in type is hashable and can be a dict key?
d = {
    "name": 1,              # str
    42: 2,                  # int
    3.14: 3,                # float
    (1, 2): 4,              # tuple (of hashables)
    frozenset({1, 2}): 5,   # frozenset
    b"bytes": 6,            # bytes
}
print(d)
```

```python
## Question 06 : Set operation chain.
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print((a ^ b) - a)
```

```python
## Question 07 : Tuple unpacking with star.
t = (10, 20, 30, 40, 50)
a, *b, c = t
print(b)
```

```python
## Question 08 : Dictionary merge (Python 3.9+).
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 4}
print(d1 | d2)
```

```python
## Question 09 : Nested conditional expression.
x, y, z = 5, 10, 15
res = "A" if x > y else ("B" if y > z else ("C" if z > x else "D"))
print(res)
```

```python
## Question 10 : filter + map pipeline.
nums = [-2, -1, 0, 1, 2, 3, 4]
filtered = filter(lambda x: x > 0, nums)
mapped = map(lambda x: x ** 2, filtered)
print(list(mapped))
```

```python
## Question 11 : functools.reduce.
from functools import reduce
vals = [1, 2, 3, 4]
res = reduce(lambda acc, x: acc - x, vals, 10)
print(res)
```

```python
## Question 12 : Mutable default argument.
def append_val(item, target=[]):
    target.append(item)
    return target

print(append_val(1))
print(append_val(2))
```

```python
## Question 13 : Nested variable scoping with `global`.
x = 10
def outer():
    x = 20
    def inner():
        global x
        x = 30
    inner()
    return x

print(outer(), x)
```

```python
## Question 14 : Encoding check.
text = "Python"
print(len(text), len(text.encode("utf-8")))
```

```python
## Question 15 : Dict comprehension with filter.
keys = ['a', 'b', 'c']
vals = [1, 2, 3]
d = {k: v**2 for k, v in zip(keys, vals) if v % 2 != 0}
print(d)
```

```python
## Question 16 : Custom sorting with tuple key.
data = [('apple', 3), ('banana', 1), ('cherry', 2)]
res = sorted(data, key=lambda x: (x[1], x[0]))
print(res[0])
```

```python
## Question 17 : Bitwise check.
x, y = 5, 3   # 0101 and 0011
if x & y and not (x | y > 10):
    print("Pass")
else:
    print("Fail")
```

```python
## Question 18 : Shallow copy.
x = [1, 2, [3, 4]]
y = list(x)
y[2].append(5)
print(x[2])
```

```python
## Question 19 : Positional indices in `str.format`.
fmt = "{1} {0} {1}"
res = fmt.format("Python", "Machine")
print(res)
```

```python
## Question 20 : Set intersection and difference.
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}
s3 = {2, 4, 6}
print((s1 & s2) - s3)
```

```python
# Question 21: Tuple immutability with a mutable element
t = (1, 2, [3, 4])
t[2].extend([5, 6])
print(t)
```

```python
## Question 22 : `setdefault`.
d = {}
d.setdefault('a', []).append(10)
d.setdefault('a', []).append(20)
print(d)
```

```python
## Question 23 : Short-circuit evaluation.
def check():
    print("Checked")
    return True

x = False
res = x and check()
print(res)
```

```python
## Question 24 : filter + reduce.
from functools import reduce
nums = [1, 2, 3, 4, 5, 6]
evens = filter(lambda x: x % 2 == 0, nums)
res = reduce(lambda a, b: a * b, evens, 1)
print(res)
```

```python
## Question 25 : Positional-only and keyword-only parameters.
def process(a, b, /, *, c, d):
    return a + b + c + d

print(process(1, 2, c=3, d=4))
```

```python
## Question 26 : `str.translate`.
trans = str.maketrans("aeiou", "12345")
msg = "machine learning"
print(msg.translate(trans))
```

```python
## Question 27 : Argument unpacking.
def compute(a, b, c, d):
    return a * b + c * d

args = (2, 3)
kws = {'c': 4, 'd': 5}
print(compute(*args, **kws))
```

```python
## Question 28 : if / elif / else.
a, b, c = 3, 7, 2
if a > b:
    res = "X"
elif b > c and a < c:
    res = "Y"
else:
    res = "Z"
print(res)
```

```python
## Question 29 : Set comprehension.
s = {x for x in range(10) if x % 2 == 0 or x % 3 == 0}
print(len(s))
```

```python
## Question 30 : Closures.
def multiplier(n):
    return lambda x: x * n

double = multiplier(2)
triple = multiplier(3)
print(double(5) + triple(4))
```

```python
## Question 31 : Float sum casting.
val = 0.1 + 0.2
print(f"{int(val * 10)}")
```

```python
## Question 32 : `popitem`.
d = {'z': 1, 'a': 2, 'm': 3}
print(d.popitem())
```

```python
## Question 33 : Transformation function.
def transform(nums):
    return list(map(lambda x: x * 2, filter(lambda x: x % 2 == 0, nums)))

print(transform(range(5)))
```

```python
## Question 34 : Nested destructuring.
data = [1, (2, 3), [4, 5]]
a, (b, c), [d, e] = data
res = a + b * c - d / e
print(int(res))
```

```python
## Question 35 : Create a tuple `Z` containing "The End" and check its type.
Z = ("The End",)   # the trailing comma makes it a tuple
print(Z)
print(type(Z))

# Without the comma it is only a string:
print(type(("The End")))
```
