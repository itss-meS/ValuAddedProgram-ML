Adding single element / iterable in list

```python
# .append() add an element to the list at the end its an in-place operation
```

```python
# list.append() method adds a single element to the end of the list. The element can be of any data type, including another list, which will be added as a single element (nested list).
l = ['s','h','r','e','y']
l.append(['u,o'])
print(l)
```

```python
# extend() method adds each element of an iterable (like a list, tuple, or string) to the end of the list. If you pass a list to extend(), it will add each element of that list individually to the original list.
l = ['s','h','r','e','y']
l.extend(['u','o'])
print(l)
```

```python
l = ['s','h','r','e','y']
new_l = l.extend([])
print(new_l)
print(len(l))
```

```python
# extend mearge multiple iterable elements in the list
a = [1]
a.extend(('hello'))
print(a)
```

```python
# .extend(destination_list_object) method adds each element of the destination_list_object to the end of the list. If you pass a list to extend(), it will add each element of that list individually to the original list.
a = [1,2,3]
b = ['a','b','c']
a.extend(b)
print(a)
print(b)
print(*a)
```

```python
#concatination
x = [1,2,3,4]
y = ['a','b']
print(x + y)
print(x)
print(y)
```

```python
# merges lists using unpacking
list1 = [1, 2, 3]
list2 = [4, 5, 6]
merged = [*list1, *list2]
print(merged)
```

```python
import itertools
list1 = [1,2,3]
list2 = [4,5,6]
merged = list(itertools.chain(list1, list2))
print(merged)
print(list1)
print(list2)
```

```python
# shallow copy
x = [1,2,3,4,5]
y = x.copy()
print(x)
print(y)
x.append(6)
print(x)
print(y)
```

```python
# deep copy of list
import copy
x = [1,2,3,4,5]
y = x
print(x)
print(y)

print(id(x))
print(id(y))
```

```python
import copy
list1 = [1, 2, 3, 4, 5]
list2 = copy.deepcopy(list1)  # Create a shallow copy of list1
print(list1)
print(list2)
print(id(list1))
print(id(list2))
```

```python
# sorted() function returns a new sorted list from the elements of any iterable.
# sorted() does not modify the original list; it creates a new list that is sorted.
nums = [5, 2, 8, 1, 9]
sorted_nums = sorted(nums)
print(nums)
print(sorted_nums)
```

```python
# sort() method sorts the list in place, modifying the original list. It does not return a new list; instead, it rearranges the elements of the existing list.
numbers = [1, 2, 3, 4, 5]
numbers.sort(reverse=True)
print(numbers)
```

```python
tuple1 = (10,'hello', 9.8)
tuple1.index('hello')
tuple1[1]
```

```python
# sorted() function can be used to sort the elements of a tuple, but it returns a new sorted list rather than modifying the original tuple. Tuples are immutable, so you cannot change their order directly.
tuple2 = (5, 2, 8, 1, 9)
sorted_tuple = sorted(tuple2)
print(sorted_tuple)
```

```python
#reverse() function can be used to reverse the elements of a tuple, but it also returns a new reversed list rather than modifying the original tuple. Tuples are immutable, so you cannot change their order directly.

tuple3 = (1, 2, 3, 4, 5)
reversed_tuple = tuple(reversed(tuple3))
print(reversed_tuple)
```

```python
#convert tuple to list
tuple4 = (1, 2, 3, 4, 5)
list_from_tuple = list(tuple4)
print(list_from_tuple)
```

```python
# sum()
tuple5 = (1, 2, 3, 4, 5)
total_sum = sum(tuple5)
print(total_sum)
```

```python
# any() and all()
tuple6 = (1, 2, 3, 4, 5)
print(any(tuple6))  # True
print(all(tuple6))  # True
```

```python
# create a dictionary
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(my_dict)
```

```python
# updating vaule
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
my_dict['age'] = 31
my_dict
```

```python
# get vaule of key
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
age_value = my_dict.get('age')
print(age_value)
```

```python
# add new key-value pair
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
my_dict['country'] = 'USA'
print(my_dict)
```

```python
# del()
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
del my_dict['age']
print(my_dict)
```

```python
#pop()
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
removed_value = my_dict.pop('age')
print(removed_value)
print(my_dict)
```

```python
dic_1 = {'a': 1, 'b': 2, 'c': 3}
dic_2 = {'b': 4, 'c': 5, 'd':6}
keyToRemove = {}
dic_1 = {k: dic_1[k] for k in dic_1 if k not in keyToRemove}
dic_2 = {k: dic_2[k] for k in dic_2 if k not in keyToRemove}
merged_dict = {**dic_1, **dic_2}
print(merged_dict)
```
