```python
# dict to array
data = {'a': 1, 'b': 2, 'c': 3}
arr = np.array(list(data.values()))
print("=="*20)
print(arr)
print("=="*20)
arr = np.array(list(data.items()))
print(arr)
print("=="*20)
arr = np.array(list(data.keys()))
print(arr)
```
